```python
from datetime import date,datetime
name = input("Enter your name: ")
dob = input("Enter a date (YYYY-MM-DD): ")
date_obj = datetime.strptime(dob, "%Y-%m-%d").date() #datetime.strptime(dob, "%Y-%m-%d")
today = date.today()
age = today.year - date_obj.year

print(f"{name} age is {age}")
```

```python
def decimal_to_binary(number):
    if number == 0:
        return '0'
    result = ''
    while number > 0:
        result += str(number % 2)
        number = number // 2
        return result[::-1]
    
a = decimal_to_binary(25.234)
print(a)
```
